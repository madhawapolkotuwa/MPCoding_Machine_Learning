# Reading CSV Files

A CSV (Comma-Separated Values) file is a plain text file that stores tabular data, with each line representing a row and commas separating the columns

````python
import pandas as pd
````

````python
# mount google drive
from google.colab import drive
drive.mount('/content/drive')
````

````python
file_path = '/content/drive/MyDrive/ML_practice_datasets/student_performance_dataset.csv'
````

````python
# Import the data
data = pd.read_csv(file_path)
````

````python
print(data.head(5))
````

````python
data.final_exam_score[:5]
````

````python
# Skip 1st row, headers(column names), and set no headers
data = pd.read_csv(file_path, header=None, skiprows=1)
data.head(3)
````

````python
# skip the 1st row and Provide custom names for the columns
data = pd.read_csv(file_path, skiprows=1 ,names=["id", "sex", "std_t", "attd_p", "sleep_h", "lvl", "web_acc", "extra", "ptj", "lst_grd", "f_scr", "grd"])
data.head(3)
````

````python
# Load only specific columns names
data = pd.read_csv(file_path, usecols=['parental_education', 'final_exam_score'])
data.head(3)
````

````python
# Check missing values
data = pd.read_csv(file_path)
# Returns the total count of NA values for each column
data.isna().sum()
````

````python
data = pd.read_csv(file_path)
print(f"len(data): {len(data)}")
# Drops the row if any column in that row contains an NA
data_clean = data.dropna()
print(f"len(data_clean): {len(data_clean)}")
````

### Different delimiters - tab-separated file (.tsv)

```py
data = pd.read_csv(file_path, sep="\t")
```

