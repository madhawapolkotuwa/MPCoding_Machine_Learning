## SQL Read

```python
import sqlite3 as sq3
import pandas.io.sql as pds
import pandas as pd
```

```python
# mount google drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
file_path = '/content/drive/MyDrive/datasets/music_playlist.db'
```

```python
con = sq3.Connection(file_path)
con
```

### Reading data

```python
# write query
query = 'SELECT * FROM music_playlist;'

# Execute query
data = pds.read_sql(query, con)
data.head()
```

```python
# write query
query = 'SELECT Song_Title FROM music_playlist;'

# Execute query
data = pds.read_sql(query, con)
data.head()
```

```python
# write query
query = 'SELECT COUNT(*) FROM music_playlist;'

# Execute query
data = pds.read_sql(query, con)
data
```

```python
# write query
query = 'SELECT DISTINCT Genre FROM music_playlist;'

# Execute query
data = pds.read_sql(query, con)
data
```

```python
query = '''
SELECT Artist, COUNT(*) AS num_songs 
    FROM music_playlist
    GROUP BY Artist
    ORDER BY num_songs desc;
'''
# Execute query
data = pds.read_sql(query, con)
data
```
